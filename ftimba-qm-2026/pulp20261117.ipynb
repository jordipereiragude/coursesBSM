I would like to formulate and solve a model in PuLP using the specified version of the library and HiGHS as the solver.

The problem has two non-negative variables (m and s).

The objective is to **MAXIMIZE**:

16*m + 10*s

**Constraints:**

2*m + 2*s <= 8

2*m + s <= 6

Start the code with:

`from pulp import *`

And remember that:

1. First, add the variables to the model (for example: `x = model.add_variable("x", 0, 4)`).

2. Then, add the objective function to the model (`model += 3*x + 2*y, "obj"`).

3. Next, add each constraint (`model += x + y <= 5, "c1"`).

Finally, solve the problem using HiGHS:

`status = model.solve(HiGHS(msg = 0))`

And print the results:

To obtain the value of a variable, use `value(x)`, where x is the variable you want to evaluate.

To obtain the value of the objective function, use `value(model.objective)`.

Use HiGHS as the solver and print the values of variables s and m, as well as the objective function value.

```python
from pulp import *

# Create the model
model = LpProblem("Maximize_Profit", LpMaximize)

# Add the variables using the requested format: model.add_variable(name, lowBound, upBound)
m = model.add_variable("m", 0, None)
s = model.add_variable("s", 0, None)

# Add the objective function to the model
model += 16 * m + 10 * s, "obj"

# Add each constraint
model += 2 * m + 2 * s <= 8, "c1"
model += 2 * m + s <= 6, "c2"

# Solve the problem using HiGHS
status = model.solve(HiGHS(msg=0))

# Print the results
print(f"m = {value(m)}")
print(f"s = {value(s)}")
print(f"Objective value = {value(model.objective)}")
```

```python
# 1. Revenue per unit for 5 products (x1, x2, x3, x4, x5)
revenue = [45, 60, 35, 80, 50]

# 2. Capacity limit for  3 resources (r1, r2, r3)
capacity = [90, 80, 95]

# 3. Use (in units) for each resource (rows) and product (columns)
# use[i][j] contains the number of units of resource i required to build a unit of product j.
# Rows: r1, r2, r3. Columns: x1, x2, x3, x4, x5.
use = [
    [15, 20, 10, 25, 12],  # Resource consumption of resource r1 for each of the 5 products
    [10, 15, 12, 30, 18],  # Resource consumption of resource r2 for each of the 5 products
    [18, 12, 15, 20, 10]   # Resource consumption of resource r3 for each of the 5 products
]
```

Extend the PuLP model to generalize the previous model using the `revenue`, `capacity`, and `use` data provided earlier.

Name this new model `m`.

```python
from pulp import *

# Create the generalized model named 'm'
m = LpProblem("Generalized_Model", LpMaximize)

# Create decision variables for the 5 products (x0 to x4)
# Using the custom method format: m.add_variable(name, lowBound, upBound)
x = [m.add_variable(f"x_{i}", 0, None) for i in range(len(revenue))]

# Define the objective function (maximize total revenue)
m += lpSum(revenue[j] * x[j] for j in range(len(revenue))), "Total_Revenue"

# Add capacity constraints for each resource
for i in range(len(capacity)):
    m += lpSum(use[i][j] * x[j] for j in range(len(revenue))) <= capacity[i], f"Capacity_Constraint_{i}"

# Solve the model using HiGHS
status = m.solve(HiGHS(msg=0))

# Print the results
for j in range(len(revenue)):
    print(f"x_{j+1} = {value(x[j])}")
print(f"Objective value = {value(m.objective)}")
```
