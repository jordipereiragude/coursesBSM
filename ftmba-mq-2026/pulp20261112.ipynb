Me gustaría formular y resolver un modelo en PuLP con la versión de la librería indicada y usando HiGHS como solver.

El problema tiene dos variables mayores e iguales a 0 ( m y s)

Hay que MAXIMIZAR 16*m+10*s

Restricciones:

2*m+2*s<=8

2*m+s<=6

Empieza el código con: from pulp import *

Y recuerda que:

1) Hay que añadir al modelo las variables (por ejemplo: x = model.add_variable("x", 0, 4))

2) Luego hay que añadir al modelo el objetivo (model += 3*x + 2*y, "obj")

3) Posteriormente cada restricción (model += x + y <= 5, "c1")

Finalmente resuelve el problema con HiGHS:

status = model.solve(HiGHS(msg = 0))


E imprime los resultados:

Para obtener el valor una variable usa: value(x) donde x es la variable que uses

Para obtener el valor de la función objetivo usa: value(model.objective)

 como solver e imprime el valor de la variable s y m, así como de la función objetivo

```python
from pulp import *

# Crear el modelo
model = LpProblem("Maximizar_Beneficio", LpMaximize)

# 1) Añadir al modelo las variables m y s >= 0
m = model.add_variable("m", 0, None)
s = model.add_variable("s", 0, None)

# 2) Añadir el objetivo
model += 16 * m + 10 * s, "obj"

# 3) Añadir las restricciones
model += 2 * m + 2 * s <= 8, "c1"
model += 2 * m + s <= 6, "c2"

# Resolver el problema con HiGHS
status = model.solve(HiGHS(msg = 0))

# Imprimir los resultados
print(f"Valor de m: {value(m)}")
print(f"Valor de s: {value(s)}")
print(f"Valor de la función objetivo: {value(model.objective)}")
```

```python
# 1. Lista de ingreso para 5 productos (x1, x2, x3, x4, x5)
ingresos = [45, 60, 35, 80, 50]

# 2. Lista de capacidades para  3 recursos (r1, r2, r3)
capacidad = [90, 80, 95]

# 3. Usoso de unidades de cada recurso (filas) por cada producto (columnas)
# matriz_uso[i][j] representa cuántas unidades del recurso i consume el producto j.
# Filas: r1, r2, r3. Columnas: x1, x2, x3, x4, x5.
matriz_uso = [
    [15, 20, 10, 25, 12],  # Consumo del recurso r1 por cada uno de los 5 productos
    [10, 15, 12, 30, 18],  # Consumo del recurso r2 por cada uno de los 5 productos
    [18, 12, 15, 20, 10]   # Consumo del recurso r3 por cada uno de los 5 productos
]
```

Extiende el modelo de PuLP para considerar generalizar el modelo anterior con los datos de ingresos, capacidad y matriz_uso indicadas anteriormente.

Llama a este nuevo modelo m

```python
from pulp import *

# Crear el modelo con el nombre 'm'
m = LpProblem("Modelo_Generalizado_Produccion", LpMaximize)

# Definir el número de productos y recursos
num_productos = len(ingresos)  # 5
num_recursos = len(capacidad)   # 3

# 1) Añadir las variables de decisión x1, x2, x3, x4, x5 al modelo m (todas >= 0)
# Usamos model.add_variable para respetar la sintaxis del primer ejercicio
productos = []
for i in range(num_productos):
    var_name = f"x{i+1}"
    var = m.add_variable(var_name, 0, None)
    productos.append(var)

# 2) Añadir la función objetivo al modelo m (Maximizar el beneficio total)
# Beneficio = Sumatoria de (ingreso_j * x_j)
m += lpSum(ingresos[j] * productos[j] for j in range(num_productos)), "obj"

# 3) Añadir las restricciones de recursos al modelo m
# Para cada recurso i: Sumatoria de (matriz_uso[i][j] * x_j) <= capacidad[i]
for i in range(num_recursos):
    m += lpSum(matriz_uso[i][j] * productos[j] for j in range(num_productos)) <= capacidad[i], f"c_{i+1}"

# Resolver el problema con HiGHS
status = m.solve(HiGHS(msg = 0))

# Imprimir los resultados de las variables y del valor óptimo
print("Estado de la solución:", status) # Muestra el código de estado retornado por el solver
for var in productos:
    print(f"Valor de {var.name}: {value(var)}")
print(f"Valor de la función objetivo: {value(m.objective)}")
```
